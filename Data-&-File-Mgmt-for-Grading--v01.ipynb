<span style=color:blue>Function to parse names</span>

```python
def parse_name(name):
    sp = name.split(' ')
    ln = sp[-1]
    fn = ''
    for i in range(0, len(sp)-1):
        fn = fn + ' ' + sp[i]
    fn = fn[1:]
    return {'ln': ln, 'fn': fn}

pprint.pp(parse_name('Richard Hull'))
pprint.pp(parse_name('Leonel Garibay-Estrada'))
pprint.pp(parse_name('Phuong Thy Nguyen'))
pprint.pp(parse_name(''))
```

<span style=color:blue>Looping through team file; separating names into rows</span>

```python
dlist = {}

for asmnt in ['PA2','PA3']:
    dlist[asmnt] = []

    file_path = dir + team_file[asmnt]
    with open(file_path, mode='r', newline='', encoding='utf-8') as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            dlist[asmnt].append(row)

    pprint.pp(dlist[asmnt])
```

```python
newlist = {}

for asmnt in ['PA2','PA3']:
 newlist[asmnt] = []
 rowcnt = 1
 for row in dlist[asmnt]:
  # pprint.pp(row)
  rowcnt = rowcnt + 1
  print (rowcnt)
  teamname = row['TeamName']
  if teamname != 'emptyTeam':
    for i in range(1,5):
        print(i)
        pn = 'Person' + str(i) + 'Name'
        em = 'Person' + str(i) + 'Email'
        # print(pn)
        # print(row[pn])
        if row[pn]!= '':
            parsed_name = parse_name(row[pn])
            newrow = {'team_number': rowcnt,
                      'team_name' : teamname,
                      'last_name' : parsed_name['ln'],
                      'first_name' : parsed_name['fn'],
                      'full_name': row[pn],
                      'email': row[em]
                     }
            pprint.pp(newrow)
            newlist[asmnt].append(newrow)
    blankrow = {'team_number': '',
                'team_name' : '',
                'last_name' : '',
                'first_name' : '',
                'full_name': '',
                'email': ''
                     }
    newlist[asmnt].append(blankrow)

 outlist[asmnt] = newlist[asmnt]
```

<span style=color:blue>Function from last name to team number     </span>

```python
def last_name_to_team_member(assignment, lname):
    personlist = []
    for row in outlist[assignment]:
        if row['last_name'] == lname:
            personlist.append(row)

    return personlist

pprint.pp(last_name_to_team_member('PA3', 'Hershman'))
pprint.pp(last_name_to_team_member('PA3', 'Qian'))
pprint.pp(last_name_to_team_member('PA2', 'Tran'))
        
    
```
